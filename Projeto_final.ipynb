Para a criação inicial das funções 

$$n!$$

```python
def funcao_fatorial(n):
    fatoriais=[1]
    for i in range(n):
        fatoriais.append(fatoriais[-1]*(i+1))
    return fatoriais
```

$$e=\sum_{n=0}^{\infty}\dfrac{1}{n!}$$

```python
def numero_de_euler(n):
    soma=0
    fatorial=funcao_fatorial(n)
    for i in range(n+1):
        y=1/fatorial[i]
        soma+=y
    return soma

euler=numero_de_euler(100)
```

Transformo um $x$ -> $x_0$ em que $x_0$ pertence ao intervalo [-1;1] na para um B em um expoente $n$, de forma padrão na função B é o número de euler.

$$\dfrac{x}{B^n}=x_0 \ \ \text{onde} \ \ |x_0|<1$$

```python
def intervalo(x,B=euler):
    x_0=x
    n=0
    while not (x_0 <= 1 and x_0 >= -1):
        x_0=x_0/B
        n+=1
    return n, x_0 
```

```python
print(intervalo(22))
print(intervalo(-20.08552))
print(intervalo(2))
print(intervalo(42))
```

<center>Função Exponencial na base de euler</center>

$$e^x=\sum_{n=0}^{\infty}\dfrac{x^n}{n!}$$

```python
def exp(expoente:float):
    #Definição de variavéis
    resultado=0
    fatoriais=funcao_fatorial(101)
    x=1
    #codigo
    for i in range(101):
        y=(x)/fatoriais[i]
        x*=expoente
        resultado+=y
    return resultado
```

```python
print(exp(75))
print(exp(1))
print(exp(0))
print(exp(-5))
```

<center>Função Logarítmica</center>

$$ln(x)=\alpha+\sum_{n=1}^{\infty}(-1)\dfrac{\left(1-\dfrac{x}{e^\alpha}\right)^n}{n}$$

```python
def ln(x:float):
    #verifica se pode ser calculado R->R
    if x < 0:
        return ValueError
    #identifica o intervalo e muda valor
    inter=intervalo(x)
    alfa=inter[0]
    x_0=inter[1]
    #valor dentro do parenteses
    x_1=(1-x_0)
    #somatorio
    ln_x=alfa
    soma=0
    for i in range(1,11):
        soma+=x_1/i
        x_1*=x_1
    #resultado
    ln_x+=-soma

    return ln_x
```

```python
print(ln(2))
print(ln(-42))
print(ln(3))
print(ln(52))
print(ln(17))    
```

Tentativa de criar o somatório de uma função genérica $f_k$

$$Soma =\sum_{k=i_0}^{i_n}f_k$$

```python
def somatorio(f_k:float,i_0=0,i_n=100):
    soma=0
    for i in range(i_0,i_n+1):
        soma+=f_k(i)
    return soma
```
