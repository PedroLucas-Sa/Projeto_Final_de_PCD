Segue a função fatorial e a criação de uma função para a criação de lista com os n+1 termos que respeitam o fatorial
$$n!$$

```python
def fat(n:int):
    """
    Input: int
    Output: fatorial do input
    """
    if n==0:
        return 1
    return fat(n-1)*n
```

```python
def lista_fatorial(n:int):
    fatoriais=[]
    for i in range(n+1):
        fatoriais.append(fat(i))
    return fatoriais
```

Realização de uma função para encontrar o número de euler
$$e=\sum_{n=0}^{\infty}\dfrac{1}{n!}$$

```python
def numero_de_euler(n):
    soma=0
    fatorial=lista_fatorial(n)
    for i in range(n+1):
        y=1/fatorial[i]
        soma+=y
    return soma

euler=numero_de_euler(100)
```

Para encontrar o valor de $\pi$ foi realizado o método rápido de Machin em que usa série de taylor da função arctg, a série de taylor será abordado mais para frente [1]
$$\pi= 4\left[4\text{arctg}(1/5) - \text{arctg}(1/239)\right]$$

```python
def arctg(x, termos):
    """Calcula o arco-tangente usando a série de Taylor"""
    soma = 0.0
    for n in range(termos):
        soma += ((-1)**n * (x**(2*n + 1))) / (2*n + 1)
    return soma
```

```python
def valor_de_pi(n):
    # Método Rápido de Machin
    pi = 4 * (4 * arctg(1/5, n) - arctg(1/239, n))
    return pi

pi=valor_de_pi(100)
```

Para algumas funções foi realizado uma analise em que transforma um $x$ -> $x_0$ em que $x_0$ pertence ao intervalo [-1;1] para um $B$ com um expoente $n$, de forma padrão na função $B$ é o número de euler.

$$\dfrac{x}{B^n}=x_0 \ \ \text{onde} \ \ |x_0|<1$$

```python
def intervalo(x:float,B=euler):
    x_0=x
    n=0
    while not (x_0 <= 1 and x_0 >= -1):
        x_0=x_0/B
        n+=1
    return n, x_0 
```

Função Logarítmica

$$ln(x)=\alpha+\sum_{n=1}^{\infty}(-1)\dfrac{\left(1-\dfrac{x}{e^\alpha}\right)^n}{n}$$

```python
def ln(x:float):
    #verifica se pode ser calculado R^+->R
    if x < 0:
        return ValueError
    #identifica o intervalo e muda valor
    inter=intervalo(x)
    alfa=inter[0]
    x_0=inter[1]
    #valor dentro do parenteses
    x_1=(1-x_0)
    #somatorio
    ln_x=alfa
    soma=0
    for i in range(1,11):
        soma+=x_1/i
        x_1*=x_1
    #resultado
    ln_x+=-soma

    return ln_x
```

Função logarítma
$$\log_{a}b=\dfrac{\ln(b)}{\ln(a)}$$

```python
def log(x:float,y:float):
    if y== 0:
        return ValueError
    return ln(x)/ln(y)
```
