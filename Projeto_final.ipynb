Segue a função fatorial e a criação de uma função para a criação de lista com os n+1 termos que respeitam o fatorial
$$n!$$

```python
def fatorial(n:int):
    """
    Input: int
    Output: fatorial do input
    """
    if n==0:
        return 1
    return fatorial(n-1)*n
```

```python
def lista_fatorial(n:int):
    """
    Recebe um valor inteiro n
    Retorna uma lista com n+1 fatoriais
    """
    fatoriais=[]
    for i in range(n+1):
        fatoriais.append(fatorial(i))
    return fatoriais
```

Realização de uma função para encontrar o número de euler
$$e=\sum_{n=0}^{\infty}\dfrac{1}{n!}$$

```python
def numero_de_euler(n:int=100):
    soma=0
    fatoriais=lista_fatorial(n)
    for i in range(n+1):
        y=1/fatoriais[i]
        soma+=y
    return soma

euler=numero_de_euler()
print(euler)
```

Para encontrar o valor de $\pi$ foi realizado o método rápido de Machin em que usa série de taylor da função arctg, a série de taylor será abordada mais para frente [2]
$$\pi= 4\left[4\cdot\text{arctg}\left(\frac{1}{5}\right) - \text{arctg}\left(\frac{1}{239}\right)\right]$$

```python
def arctg(x: float, termos: int=10):
    """Calcula o arco-tangente usando a série de Taylor"""
    soma = 0.0
    for n in range(termos):
        soma += (pow(-1,n) * (pow(x,(2*n + 1)))) / (2*n + 1)
    return soma
```

```python
def valor_de_pi(n: int=100):
    """Calcula o valor de pi usando o método rápido de Machin"""
    # Método Rápido de Machin
    pi = 4 * (4 * arctg(1/5, n) - arctg(1/239, n))
    return pi

pi=valor_de_pi()
print(pi)
```

Para algumas funções foi realizado uma analise em que transforma um $x$ -> $x_0$ em que $x_0$ pertence ao intervalo [-1;1] para um $B$ com um expoente $n$, de forma padrão na função $B$ é o número de euler.

$$\dfrac{x}{B^n}=x_0 \ \ \text{onde} \ \ |x_0|<1$$

```python
def intervalo(x:float,B: float =euler):
    x_0=x
    n=0
    while not (x_0 <= 1 and x_0 >= -1):
        x_0=x_0/B
        n+=1
    return n, x_0 
```

#### Função Ln e Log

A função ln(x) foi implementada usando uma série de Taylor com mudança de variável para aproximar valores fora do intervalo de convergência direto. Primeiro, o valor x é transformado para um x₀ em [-1,1], depois a série é somada para obter ln(x).

$$\ln(x)=\ln(e^\alpha)+\sum_{n=1}^{\infty}(-1)\dfrac{\left(1-\dfrac{x}{e^\alpha}\right)^n}{n}$$

```python
def ln(x:float):
    #verifica se pode ser calculado R^+->R
    if x <= 0:
        return ValueError
    #identifica o intervalo e muda valor
    inter=intervalo(x)
    alfa=inter[0]
    x_0=inter[1]
    #valor dentro do parenteses
    x_1=(1-x_0)
    #somatorio
    ln_x=alfa
    soma=0
    for i in range(1,100):
        x_2=pow(x_1,i)
        soma+=x_2/i
    #resultado
    ln_x+=-soma

    return ln_x
```

Função logarítma foi realizada por meio da função $\ln(x)$

$$\log_{y}x=\dfrac{\ln(x)}{\ln(y)}$$

```python
def log(x: float, y: float):
    if x <= 0 or y <= 0 or y == 1:
        raise ValueError("x e y devem ser positivos e y não pode ser 1")
    return ln(x) / ln(y)
```

#### Função Seno

A função seno foi aproximada pela série de Taylor para valores em radianos. Para melhorar a convergência, o ângulo é reduzido para o intervalo $[-\frac{\pi}{2}, \frac{\pi}{2}]$.

$$\text{sen}(x)=\sum_{n=0}^{\infty}(-1)^{n}\dfrac{x^{2n+1}}{(2n+1)!}$$

```python
def intervalo_angulo(x:float):
    """
    Recebe um valor x
    Transforma o valor x para um valor correspondente dado no intervalo [-pi/2,pi/2]
    Retorna esse valor transformado
    """
    while x>pi:
        x=x-2*pi
    while x<-pi:
        x=x+2*pi
    if x>pi/2:
        x=pi-x
    elif x<-pi/2:
        x=-pi+abs(x)
    return x
```

```python
def sen(x:float,rad: bool=True):
    """
    Recebe um valor em ângulo, se o ângulo for em graus necessecita colocar em rad False
    Retorna o seno desse ângulo
    """
    #analise se é radiano ou graus
    if rad == False:
        x=x*pi/180
    #transformação do ângulo
    x=intervalo_angulo(x)
    #variaveis
    soma=0
    k=-1
    lista=lista_fatorial(101)
    for i in range(50):
        k*=-1
        c=(2*i+1)
        a=pow(x,c)
        b=lista[c]
        soma+=k*(a/b)
    if soma>1 and soma<1.2:
        soma=1.0
    elif soma<-1 and soma>-1.2:
        soma=-1.0
    if soma<=1 and soma>=-1:
        return soma
    else:
        return print("analise o input")
```

#### Função Cosseno
Para a função cosseno foi utilizada a relação trigonométrica:
$$\cos(x)=\sin\left(x+\frac{\pi}{2}\right)$$
Como já foi progamada a função seno e existe essa relação, a progamação do cosseno utilizou a função seno como base

```python
def cos(x:float,rad: bool= True):
    """
    Recebe um valor em ângulo, se o ângulo for em graus necessecita colocar em rad False
    Retorna o cosseno desse ângulo
    """
    a=x+pi/2
    return sen(a,rad)
```

#### Função Tangente
Assim como a função cosseno a função tangente pode ser descrita como uma função de seno e cosseno, a função tangente foi implementada utilizando elas desta maneira

$$\tg(x)=\frac{\sin(x)}{\cos(x)}$$

```python
def tg(x:float,rad: bool= True):
    """
    Recebe um valor em ângulo, se o ângulo for em graus necessecita colocar em rad False
    Retorna a tangente desse ângulo
    """
    seno=sen(x,rad)
    cosseno=cos(x,rad)
    if cosseno==0:
        return ZeroDivisionError
    return seno/cosseno
```

#### Função Arco-seno
A função arco-cosseno foi implementada utilizando a relação com o arco-seno. Esta função retorna o ângulo cujo cosseno é o valor $x$ fornecido. O resultado está sempre entre $0$ e $\pi$ radianos.

$$\arcsin(x) = \sum_{n=0}^{\infty} \frac{(2n)! \ x^{2n+1}}{4^n (n!)^2 (2n+1)}$$

```python
def arcsen(x: float, termos: int = 10):
    """Calcula o arco seno usando série de Taylor"""
    if x < -1 or x > 1:
        return ValueError
    
    soma = 0.0
    for n in range(termos):
        num = 1
        den = 1
        
        # calcula (2n)!
        for i in range(1, 2*n + 1):
            num *= i
        
        # calcula (n!)^2
        for i in range(1, n + 1):
            den *= i*i
        
        coef = num / ( (2**(2*n)) * den * (2*n + 1) )
        
        soma += coef * pow(x, (2*n + 1))
    
    return soma
```

#### Função Arco-cosseno
A função arco-cosseno foi implementada utilizando a relação com o arco-seno. Esta função retorna o ângulo cujo cosseno é o valor $x$ fornecido. O resultado está sempre entre $0$ e $\pi$ radianos.

$$\arccos(x) = \frac{\pi}{2} - \arcsin(x)$$

```python
def arccos(x: float, termos: int = 10):
    """Calcula o arco cosseno usando a função arcsen"""
    if x < -1 or x > 1:
        return ValueError
    
    return (pi/2) - arcsen(x, termos)
```

Criação do somatório de uma função genérica $f(i,x)$

$$\sum_{k=i_0}^{i_n}f(k,x)=f(i_0,x)+\cdots+f(i_n,x)$$

```python
def somatorio(f, i_0: int =0, i_n: int= 100, x: float= 0):
    """
    Realiza o Somattorio de uma função f
    f tem dois inpunts o passo e um valor constante
    Retorna a soma
    """
    soma=0
    for i in range(i_0,i_n+1):
        soma+=f(i,x)
    return soma
```

### Funções Básicas Matemáticas

As funções abaixo implementam operações matemáticas elementares que serão utilizadas posteriormente na construção de funções mais complexas. Cada função recebe um valor `x` e retorna o resultado da operação aplicada.

- **Identidade:** $f(x) = x$
- **Constante:** $f(x) = c$
- **Afim:** $f(x) = ax + b$
- **Quadrática:** $f(x) = ax^2 + bx + c$
- **Modular:** $f(x) = |x|$
- **Potência:** $f(x) = a_n \cdot x^n$
- **Raiz:** $f(x) = a_n \cdot x^{1/n}$
- **Polinomial:** $f(x) = a_0 + a_1x + a_2x^2 + ...$
- **Racional:** $f(x) = \frac{P(x)}{Q(x)}$
- **Exponencial:** $f(x) = a^x$
- **e elevado a x:** $f(x) = e^x$

```python
def funcao_identidade(x:float):
    """
    O input é um valor float
    O return o mesmo valor
    """
    return x

def funcao_constante(x:float,c: float=0.0):
    """
    O input é um valor float e uma constante float
    O return é a constante
    """
    return c

def funcao_afim(x:float,a: float=1.0,b: float=0.0):
    """
    O input é um valor x float e duas constante float
    O return é o valor da função afim em x
    """
    return a*x+b

def funcao_quadratica(x:float,a:float=1,b:float=0,c:float=0):
    """
    O input é um valor x float e três constante float
    O return é o valor da função quadrática em x
    """
    resultado=a*(x**2)+b*x+c
    return resultado

def funcao_modular(x:float):
    """
    O input é um valor x float
    O return é modulo desse valor
    """
    resultado=abs(x)
    return resultado

def funcao_potencia(x:float,n:float=2,a_n:float=1):
    """
    O input é um valor x float, n o expoente, a_n a constante que acompanha
    O return é esse valor elevado a n vezes uma constante a_n
    """
    resultado=a_n*pow(x,n)
    return resultado

def funcao_raiz(x:float,n:float=2,a_n:float=1):
    """
    O input é um valor x float, n é raiz-enésima, a_n a constante que acompanha
    O return é esse valor elevado a n vezes uma constante a_n
    """
    if n==0:
        return ZeroDivisionError
    m=1/n
    return a_n*pow(x,m)

def funcao_polinomial(x:float,termos:list):
    """
    O input é um valor x float, e uma lista com a_n com constantes que acompanham cada termo
    O return é esse x elevado a n uma constante a_n para cada elemento da lista
    """
    resultado=0
    for i in range(len(termos)):
        resultado+=termos[i]*pow(x,i)
    return resultado

def funcao_racional(x_p:float,termos_p:list,x_q:float,termos_q:list,P=funcao_polinomial,Q=funcao_polinomial):
    """
    O input é um valor x float, e uma lista com termos com constantes que acompanham cada termo, isso para a função do denominador e do númerador
    O return é esse x elevado a n uma constante a_n para cada elemento da lista
    """
    Polinomio_p=P(x_p,termos_p)
    Polinomio_q=Q(x_q,termos_q)
    #detecta se vai ter divisão por zero
    if Polinomio_q==0:
        return ZeroDivisionError
    resultado=Polinomio_p/Polinomio_q
    return resultado

def funcao_exponencial(x:float,a:float):
    """
    O input é um valor x float, e um valor float a em que é a base
    O return é esse a elevado x
    """
    resultado=pow(a,x)
    return resultado

def exp(expoente:float):
    """
    O input é um valor expoente float
    O return é euler elevado ao expoente
    """
    #Definição de variavéis
    resultado=pow(euler,expoente)
    return resultado
```

### Construtor de Funções Definidas por Partes

Esta função permite criar uma função definida por partes interativamente. O usuário pode combinar diferentes funções matemáticas em intervalos distintos, especificando:

- Quantas partes a função terá
- Qual função usar em cada parte
- O intervalo de valores para cada parte
- A granularidade (número de divisões) para calcular pontos

```python
def criar_funcao_por_partes():
    """
    Cria uma função definida por partes em que se quantas funções você quer calcular
    Mostra as funções disponiveis
    Pede o intervalo para os valores da função que é em formato de lista
    E a divisão de cada intervalo
    Nessecita ser salva
    """
    dicionario_de_funcoes = {
    1:  [funcao_identidade, "y = x"],
    2:  [funcao_constante, "y = c"],
    3:  [funcao_afim, "y = ax + b"],
    4:  [funcao_quadratica, "y = ax^2 + bx + c"],
    5:  [funcao_modular, "y = |x|"],
    6:  [funcao_potencia, "y = a_n * x^n"],
    7:  [funcao_raiz, "y = a_n * x^(1/n)"],
    8:  [funcao_polinomial, "y = a0 + a1*x + a2*x^2 + ..."],
    9:  [funcao_racional, "y = P(x) / Q(x)"],
    10: [funcao_exponencial, "y = a^x"],
    11: [exp, "y = e^x"],
    12: [sen, "y = sen(x)"],
    13: [cos, "y = cos(x)"],
    14: [tg, "y = tg(x)"],
    15: [ln, "y = ln(x)"],
    16: [log, "y = log_y(x)"],
    17: [arcsen, "y = arcsen(x)"],
    18: [arccos, "y = arccos(x)"],
    19: [arctg, "y = arctg(x)"]
    }
    funcoes={
        
    }
    i=int(input("Quantas partes"))
    print("Funções:")
    for j in dicionario_de_funcoes:
        print(f'{j}: {dicionario_de_funcoes[j][1]}')
    
    for k in range(i):
        a="função "+str(k+1)+" de "+str(i)
        num = int(input("O número dessa função: "))
        b = dicionario_de_funcoes[num]
        intervalo_str = input("Intervalo neste formato [a,b]: ")
        c = list(map(float, intervalo_str.strip("[]").split(",")))
        d = int(input("Divisão de cada intervalo (1, 10, 100...): "))
        parametros = {}

        # cada função pede seus parâmetros específicos
        if num == 2:  # constante
            parametros["c"] = float(input("Digite o valor de c: "))

        elif num == 3:  # afim
            parametros["a"] = float(input("Digiteo valor de a: "))
            parametros["b"] = float(input("Digiteo valor de b: "))

        elif num == 4:  # quadrática
            parametros["a"] = float(input("Digiteo valor de a: "))
            parametros["b"] = float(input("Digiteo valor de b: "))
            parametros["c"] = float(input("Digiteo valor de c: "))

        elif num == 6:  # potência
            parametros["n"] = float(input("Digite o expoente n: "))
            parametros["a_n"] = float(input("Digite a constante a_n: "))

        elif num == 7:  # raiz
            parametros["n"] = float(input("Digite o índice da raiz n: "))
            parametros["a_n"] = float(input("Digite a constante a_n: "))

        elif num == 8:  # polinomial
            termos = input("Digite os coeficientes separados por vírgula: ")
            parametros["termos"] = list(map(float, termos.split(",")))

        elif num == 9:  # racional
            termos_p = input("Coeficientes do numerador: ")
            termos_q = input("Coeficientes do denominador: ")
            parametros["termos_p"] = list(map(float, termos_p.split(",")))
            parametros["termos_q"] = list(map(float, termos_q.split(",")))

        elif num == 10:  # exponencial
            parametros["a"] = float(input("Digite a base a: "))

        elif num in [12, 13]:  # seno e cosseno
            parametros["rad"] = bool(int(input("Em radianos? (1=sim, 0=não): ")))
        
        elif num == 14:  # tangente
            parametros["rad"] = bool(int(input("Em radianos? (1=sim, 0=não): ")))      

        elif num == 16:  # log
            parametros["base"] = float(input("Digite a base do log: "))

        elif num in [17, 18, 19]:  # arcsen, arccos, arctg
            parametros["termos"] = int(input("Digite o número de termos da série (ex: 10, 20): "))

        # salvamento
        funcoes[a] = {
            "funcao": b[0],
            "descricao": b[1],
            "intervalo": c,
            "divisoes": d,
            "parametros": parametros
        }
        
    return funcoes        
```

### Calculadora de Valores para Funções por Partes

Esta função recebe um dicionário com funções definidas por partes e calcula os valores de $y$ para cada $x$ no intervalo especificado. Retorna uma lista de pares ordenados $(x, y)$ ou $(nome, x, y)$ dependendo do parâmetro `nome`.

```python
def calculo_funcao_definida_por_partes(D: dict, nome: bool = False):
    """
    Recebe um dicionário de funções por partes
    Retorna lista de pares ordenados (x, y) ou (nome, x, y)
    """
    valores = []

    for chave, j in D.items():
        f = j["funcao"]
        a = float(j["intervalo"][0])
        b = float(j["intervalo"][1])
        d = j["divisoes"]
        parametros = j["parametros"]

        passo=(b-a)/d

        x=a
        while x <= b:
            try:
                y = f(x, **parametros)
            except TypeError:
                y = f(x)

            if nome:
                valores.append([chave, round(x, 5), round(y, 5)])
            else:
                valores.append([round(x, 5), round(y, 5)])

            x += passo

    return valores
```

### Operações de Cálculo: Derivada e Integral Numéricas

As funções abaixo implementam aproximações numéricas para:

- **Derivada:** Calcula a taxa de variação usando diferenças finitas: $\frac{dy}{dx} = \frac{y_{i+1} - y_i}{x_{i+1} - x_i}$
- **Integral:** Calcula a área sob a curva usando o método dos trapézios: $\int_a^b f(x)dx \approx \sum \frac{(y_i + y_{i+1})}{2} \cdot \Delta x$

```python
def derivada(valores, nome: bool = False):
    """
    Recebe lista de valores da função
    Recomendado utilizar uma função
    Retorna derivada numérica
    """
    deriv = []

    for i in range(len(valores) - 1):
        if nome:
            _, x1, y1 = valores[i]
            _, x2, y2 = valores[i + 1]
        else:
            x1, y1 = valores[i]
            x2, y2 = valores[i + 1]

        dx = x2 - x1
        dy = y2 - y1

        if dx == 0:
            continue

        d = dy / dx

        if nome:
            deriv.append([valores[i][0], round(x1, 5), round(d, 5)])
        else:
            deriv.append([round(x1, 5), round(d, 5)])

    return deriv

def integral(valores, nome: bool = False):
    """
    Recebe lista de valores da função
    Recomendado utilizar uma função
    Retorna integral númerica do inicio do intervalo até o final dele
    """
    integ = []
    soma = 0

    for i in range(len(valores) - 1):
        if nome:
            _, x1, y1 = valores[i]
            _, x2, y2 = valores[i + 1]
        else:
            x1, y1 = valores[i]
            x2, y2 = valores[i + 1]

        dx = x2 - x1

        area = (y1 + y2) * dx / 2
        soma += area

        if nome:
            integ.append([valores[i][0], round(x2, 5), round(soma, 5)])
        else:
            integ.append([round(x2, 5), round(soma, 5)])

    return integ
```
