Para a criação inicial das funções 

$$n!$$

```python
def funcao_fatorial(n):
    fatoriais=[1]
    for i in range(n):
        fatoriais.append(fatoriais[-1]*(i+1))
    return fatoriais
```

$$e=\sum_{n=0}^{\infty}\dfrac{1}{n!}$$

```python
def numero_de_euler(n):
    soma=0
    fatorial=funcao_fatorial(n)
    for i in range(n+1):
        y=1/fatorial[i]
        soma+=y
    return soma

euler=numero_de_euler(100)
```

<center>Intervalo de 0 a 1 na base B de forma primaria B é o número de euler</center>

$$n \Rightarrow x_0=\dfrac{x}{B^n} \ \ \text{onde} \ \ |x_0|<1$$

```python
def intervalo(n,base=euler):
    N=n
    contador=0
    while not (N <= 1 and N >= -1):
        N=N/base
        contador+=1
    return contador, N 
```

<center>Função Logarítmica</center>

$$ln(x)=\alpha+\sum_{n=1}^{\infty}(-1)\dfrac{\left(1-\dfrac{x}{e^\alpha}\right)^n}{n}$$

```python
def ln(x):
    #verifica se pode ser calculado R->R
    if x < 0:
        return ValueError
    #identifica o intervalo e muda valor
    inter=intervalo(x)
    a=inter[0]
    x_0=inter[1]
    #valor dentro do parenteses
    x_1=(1-(x_0))
    #somatorio
    n=1
    while n<100:
        ln_x=a
        ln_x-=x_1/n
        x_1*=x_1
        n+=1

    return ln_x
```
